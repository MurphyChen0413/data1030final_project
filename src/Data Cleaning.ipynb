## Data Cleaning

```python
# start with keeping the necessary columns
columns_to_keep = [
    "Maker",
    "Genmodel",
    "Color",
    "Reg_year",
    "Bodytype",
    "Runned_Miles",
    "Engin_size",
    "Gearbox",
    "Fuel_type",
    "Price",
    "Seat_num",
    "Door_num",
]

df = df[columns_to_keep].copy()

# check the new data columns
print(df.shape)
print(df.columns.tolist())
```

```python
# clean and standardize the names
df.columns = (
    df.columns
    .str.strip()            
    .str.lower()             
    .str.replace(" ", "_")     
    .str.replace("-", "_") 
)

# check the cleaned column names
print(df.columns.tolist())

# rename some columns for better clarity
df = df.rename(columns={
    "runned_miles": "mileage",
    "engin_size": "engine_size",
    "genmodel": "model",
    "maker": "brand"
})

# check the renamed column names
print(df.columns.tolist())
```

```python
# check data types
print(df.dtypes)
```

```python
# clean numeric columns
def clean_numeric(series):
    s = series.astype(str).str.lower()
    s = s.str.replace(",", "", regex=False)
    s = s.str.replace("km", "", regex=False)
    s = s.str.replace("l", "", regex=False)
    return pd.to_numeric(s.str.extract(r"(-?\d+\.?\d*)")[0], errors="coerce")

# apply the cleaning function to numeric columns
df["price"] = clean_numeric(df["price"])
df["mileage"] = clean_numeric(df["mileage"])
df["engine_size"] = clean_numeric(df["engine_size"])

for col in ["reg_year", "seat_num", "door_num"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    
# create car age column based on registration year
df["reg_year"] = df["reg_year"].astype("Int64")
df["car_age"] = 2025 - df["reg_year"]

# check cleaned numeric columns
print(df[["price", "mileage", "engine_size", "reg_year", "car_age", "seat_num", "door_num"]].head())
```

```python
# clean and standardize fuel type
def simplify_fuel_type(x):
    if pd.isna(x):
        return "Other"
    x = x.lower().strip()
    if "petrol" in x and "electric" in x:
        return "Hybrid"
    elif "diesel" in x and "electric" in x:
        return "Hybrid"
    elif "plug" in x:
        return "Hybrid"
    elif "hybrid" in x:
        return "Hybrid"
    elif "electric" in x:
        return "Electric"
    elif "petrol" in x:
        return "Petrol"
    elif "diesel" in x:
        return "Diesel"
    else:
        return "Other"

df["fuel_type_clean"] = df["fuel_type"].apply(simplify_fuel_type)
```

```python
# clean and standardize body type
def simplify_bodytype(x):
    if pd.isna(x):
        return "Other"
    x = x.lower().strip()
    if "suv" in x:
        return "SUV"
    elif "hatch" in x:
        return "Hatchback"
    elif "saloon" in x:
        return "Saloon"
    elif "estate" in x:
        return "Estate"
    elif "convertible" in x:
        return "Convertible"
    elif "coupe" in x:
        return "Coupe"
    elif "mpv" in x:
        return "MPV"
    else:
        return "Other"

df["bodytype_clean"] = df["bodytype"].apply(simplify_bodytype)
```

```python
# clean and standardize color
def clean_color(x):
    if pd.isna(x):
        return "Other"
    x = x.lower().strip()
    if "black" in x:
        return "Black"
    elif "white" in x:
        return "White"
    elif "grey" in x or "gray" in x or "silver" in x:
        return "Grey/Silver"
    elif "blue" in x or "navy" in x or "turquoise" in x or "indigo" in x:
        return "Blue"
    elif "red" in x or "maroon" in x or "burgundy" in x or "magenta" in x or "pink" in x:
        return "Red/Purple"
    elif "green" in x:
        return "Green"
    elif "yellow" in x or "gold" in x or "beige" in x or "bronze" in x:
        return "Yellow/Gold"
    elif "brown" in x:
        return "Brown"
    elif "orange" in x:
        return "Orange"
    elif "multi" in x:
        return "Multicolour"
    elif "camper" in x:  # sometimes odd vehicle label ends up in color
        return "Other"
    else:
        return "Other"

df["color_clean"] = df["color"].apply(clean_color)
```

```python
# combine car brand and model into a single feature
df["brand_model"] = df["brand"].astype(str) + " " + df["model"].astype(str)

# drop the original columns
df = df.drop(columns=["brand", "model"], errors="ignore")

# drop rows with missing target variable 'price'
df = df[df["price"].notna()]

cols_to_drop = [
    "color",
    "bodytype",
    "fuel_type",
    "brand",
    "model",
    "reg_year"
]

df = df.drop(columns=cols_to_drop, errors="ignore")

print(df.dtypes)
```
